```python
from pathlib import Path
import tifffile
import matplotlib.pyplot as plt

image_dir = Path("../data/raw/bbbc039/images/images")
image_paths = sorted(image_dir.glob("*.tif"))

print("Number of images:", len(image_paths))
```

```python
fig, axes = plt.subplots(2,5, figsize = (16,7))
for ax, path in zip(axes.flat, image_paths[:10]):
    image = tifffile.imread(path)

    ax.imshow(image, cmap = "gray")
    ax.set_title(path.name[:12], fontsize = 9)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
image = tifffile.imread(image_paths[0])

print("Shape:", image.shape)
print("Data type:", image.dtype)
print("Min:", image.min())
print("Max:", image.max())
print("Mean:", image.mean())
```

```python
mean_intensities = []
for path in image_paths[:10]:
    image = tifffile.imread(path)
    mean_intensities.append(image.mean())

plt.figure(figsize= (10,4))
plt.bar(range(1,11), mean_intensities)
plt.xlabel("Image Number")
plt.ylabel("Mean Pixcel Intensity")
plt.title("Mean Intensity of 10 Mircoscopy Images")
plt.show()
```
